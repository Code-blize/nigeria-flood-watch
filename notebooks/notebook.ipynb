```python
import zipfile
import os

# Replace with your actual zip filenames
zip_files = ['lokoja_lga_utm32n.zip', 'EMSR314_02LOKOJA_01DELINEATION_MAP_v1_vector.zip']

for z_file in zip_files:
    if os.path.exists(z_file):
        # Create a folder name based on the zip file (e.g., 'data')
        extract_folder = os.path.splitext(z_file)[0] 
        with zipfile.ZipFile(z_file, 'r') as zip_ref:
            zip_ref.extractall(extract_folder)
        print(f"Extracted {z_file} into folder: {extract_folder}")
    else:
        print(f"File not found: {z_file}")
```
